```python
from unstructured.partition.pdf import partition_pdf

file_path = "content/attention.pdf"

chunks = partition_pdf(
    filename=file_path,
    strategy="hi_res",
    infer_table_structure=True,
    extract_image_block_types=["Image"],
    extract_image_block_to_payload=True,
    chunking_strategy="by_title",
    max_characters=10000,
    combine_text_under_n_chars=2000,
    new_after_n_chars=6000,
)

print(f"Extracted {len(chunks)} chunks")
```

```python
# Get the images from the CompositeElement objects
def get_images_base64(chunks):
    images_b64 = []
    for chunk in chunks:
        if "CompositeElement" in str(type(chunk)):
            chunk_els = chunk.metadata.orig_elements
            for el in chunk_els:
                if "Image" in str(type(el)):
                    images_b64.append(el.metadata.image_base64)
    return images_b64

images = get_images_base64(chunks)
```

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

prompt = ChatPromptTemplate.from_template("""
Summarize the following excerpt concisely, preserving its
main ideas, technical details, names, and important numbers.
Return only the summary.

Excerpt:
{element}
""")

summarize_chain = (
    {"element": lambda x: x}
    | prompt
    | model
    | StrOutputParser()
)
```

```python
print(summarize_chain.invoke("Summarize this: Transformers use self-attention."))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100,
    separators=["\n\n", "\n", ". ", " ", ""]
)
```

```python
import time
from groq import APIStatusError, RateLimitError

def summarize_long_item(item):
    content = item.text if hasattr(item, "text") else str(item)
    content = content.strip()

    if not content:
        return ""

    parts = splitter.split_text(content)
    summaries = []

    for part in parts:
        for attempt in range(5):
            try:
                summary = summarize_chain.invoke(part)
                summaries.append(summary)
                break

            except RateLimitError as e:
                if attempt == 4:
                    raise

                # Respect Retry-After if Groq supplies it
                retry_after = e.response.headers.get("retry-after")
                wait = float(retry_after) if retry_after else 10 * (attempt + 1)
                print(f"Rate limited. Retrying in {wait}s...")
                time.sleep(wait)

            except APIStatusError as e:
                if e.status_code >= 500 and attempt < 4:
                    wait = 2 ** (attempt + 1)
                    print(f"Groq server error. Retrying in {wait}s...")
                    time.sleep(wait)
                else:
                    raise

    # Combine part summaries into one summary
    combined = "\n".join(summaries)

    if len(summaries) == 1:
        return summaries[0]

    return summarize_chain.invoke(combined)
```

```python
print(summarize_long_item(text[0]))
```

```python
text_summaries = [""] * len(text)

for i, item in enumerate(text):
    print(f"Processing {i + 1}/{len(text)}")

    try:
        text_summaries[i] = summarize_long_item(item)
        print("✅ Done")
    except Exception as e:
        print(f"❌ Chunk {i} failed: {e}")
        break

print("Successful:", sum(bool(s.strip()) for s in text_summaries))
```

```python
import os
import time
import base64
from google import genai
from google.genai import types

# Set your Gemini API key
# In Colab, you can enter it securely using getpass:
from getpass import getpass
os.environ["GEMINI_API_KEY"] = getpass("Gemini API key: ")

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

prompt_template = """
Describe this image from the research paper "Attention Is All You Need".
Explain the architecture, labels, arrows, graphs, and important results.
Be specific, and do not invent details.
"""

image_summaries = []

for i, image in enumerate(images):
    print(f"Summarizing image {i + 1}/{len(images)}")

    try:
        image_bytes = base64.b64decode(image)

        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=[
                prompt_template,
                types.Part.from_bytes(
                    data=image_bytes,
                    mime_type="image/jpeg"
                )
            ]
        )

        image_summaries.append(response.text or "")
        time.sleep(2)  # Gentle pacing for free-tier limits

    except Exception as e:
        print(f"Image {i} failed: {e}")
        image_summaries.append("")

print("Image summaries generated:", sum(bool(s) for s in image_summaries))
```

```python
import uuid

from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_core.stores import InMemoryStore
from langchain_classic.retrievers.multi_vector import MultiVectorRetriever
from langchain_huggingface import HuggingFaceEmbeddings
```

```python

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = Chroma(
    collection_name="multi_modal_rag",
    embedding_function=embedding_model
)

store = InMemoryStore()
id_key = "doc_id"

retriever = MultiVectorRetriever(
   vectorstore= vectorstore,
   docstore = store,
   id_key = id_key
)
```

```python
import uuid
from langchain_core.documents import Document

# 1. Text chunks
doc_ids = [str(uuid.uuid4()) for _ in text]
summary_docs = [
    Document(page_content=s, metadata={id_key: doc_ids[i], "type": "text"})
    for i, s in enumerate(text_summaries) if s.strip()
]
retriever.vectorstore.add_documents(summary_docs)
retriever.docstore.mset(list(zip(doc_ids, [t.text for t in text])))

# 2. Image chunks
image_ids = [str(uuid.uuid4()) for _ in images]
image_docs = [
    Document(page_content=s, metadata={id_key: image_ids[i], "type": "image"})
    for i, s in enumerate(image_summaries) if s.strip()
]
retriever.vectorstore.add_documents(image_docs)
retriever.docstore.mset(list(zip(image_ids, images)))

# 3. Table chunks
table_ids = [str(uuid.uuid4()) for _ in tables]
table_docs = [
    Document(page_content=s, metadata={id_key: table_ids[i], "type": "table"})
    for i, s in enumerate(table_summaries) if s.strip()
]
retriever.vectorstore.add_documents(table_docs)
retriever.docstore.mset(list(zip(table_ids, [t.text for t in tables])))
```
